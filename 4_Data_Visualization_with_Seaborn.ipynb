```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

mpg = pd.DataFrame({
    "weight": [1800, 2200, 2500, 3000, 3500, 4000],
    "horsepower": [90, 110, 130, 160, 190, 220]
})

g = sns.relplot(
    x="weight",
    y="horsepower",
    data=mpg,
    kind="scatter"
)

g.fig.suptitle("Car Weight vs. Horsepower")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

mpg_mean = pd.DataFrame({
    "model_year": [70, 71, 72, 73, 74, 75],
    "mpg_mean": [18.0, 19.5, 20.2, 21.5, 22.8, 24.0],
    "origin": [
        "USA", "Europe", "Japan",
        "USA", "Europe", "Japan"
    ]
})

g = sns.lineplot(
    x="model_year",
    y="mpg_mean",
    data=mpg_mean,
    hue="origin"
)

g.set_title("Average MPG Over Time")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create DataFrame
survey_data = pd.DataFrame({
    "Gender": [
        "Male", "Male", "Male", "Male",
        "Female", "Female", "Female", "Female"
    ],
    "Age": [18, 20, 22, 24, 17, 19, 21, 23],
    "Interested in Pets": [
        "Yes", "No", "Yes", "No",
        "Yes", "No", "Yes", "No"
    ]
})

# Set palette to "Blues"
sns.set_palette("Blues")

# Create box plot with subgroups
g = sns.catplot(
    x="Gender",
    y="Age",
    data=survey_data,
    kind="box",
    hue="Interested in Pets"
)

# Set title
g.figure.suptitle("Age of Those Interested in Pets vs. Not")

# Show plot
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create DataFrame
survey_data = pd.DataFrame({
    "Village - town": [
        "Village", "Town", "Village", "Town",
        "Village", "Town", "Village", "Town",
        "Village", "Town"
    ],
    "Likes Techno": [1, 0, 1, 1, 0, 1, 1, 0, 1, 0],
    "Gender": [
        "Male", "Male", "Male", "Male", "Male",
        "Female", "Female", "Female", "Female", "Female"
    ]
})

# Set style
sns.set_style("dark")

# Create red bar plot
g = sns.catplot(
    x="Village - town",
    y="Likes Techno",
    data=survey_data,
    kind="bar",
    col="Gender",
    color="red"
)

# Add title
g.figure.suptitle(
    "Percentage of Young People Who Like Techno",
    y=1.02
)

# Set axis labels
g.set(
    xlabel="Location of Residence",
    ylabel="% Who Like Techno"
)

plt.show()
```
