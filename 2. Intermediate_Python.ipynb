**Pandas** (Pandas Commands)


*   Import pandas > import pandas as pd

*   Create DataFrame >  pd.DataFrame()

* Read CSV > pd.read_csv()

* First rows > df.head()

* Column names >> df.columns

* Data types >> df.dtypes

* Summary info  >> df.info()

* Statistics >> df.describe()

* Select column > df['col']

* Filter rows >> df[df['col'] > x]

* Sort >> df.sort_values()

* Save CSV >  df.to_csv()



```python
import pandas as pd

data = {
    "ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emu", "Sifat"],
    "Age": [22, 25, 30, 28, 24, 26],
    "City": ["Dhaka", "Chattogram", "Khulna", "Rajshahi", "Sylhet", "Mymensingh"],
    "Salary": [30000, 45000, 50000, 40000, 35000, 50000]
}

df = pd.DataFrame(data)

print(df)

#Shows the first 5 rows.
print(df.head())

#Shows the last 5 rows.
print(df.tail())

#Returns (rows, columns).
print(df.shape)

#Displays column names.
print(df.columns)

#Shows data types and missing values.
df.info()

#Shows statistics.
print(df.describe())

#Select One Column
print(df["Name"])

print(df[["Name", "Salary"]])

#Rows
print(df.iloc[0])
print(df.iloc[1:4])
print(df.iloc[:,2])

#Filtering
print(df[df["Age"]>25])

#Max, min , sum
print(df["Salary"].max())
print(df["Salary"].min())
print(df["Salary"].sum())

#Add New Column
df["Bonus"]=[5000,6000,7000,5000,4000, 8000]

print(df)


#Sort Values
#ascending
print(df.sort_values("Salary"))
#descending
print(df.sort_values("Salary",ascending=False))
```

**Logic, Control Flow and Filtering**


```python
# Define variables
my_kitchen = 18.0
your_kitchen = 14.0

# my_kitchen bigger than 10 and smaller than 18?
print(my_kitchen >10 and my_kitchen< 18)

# my_kitchen smaller than 14 or bigger than 17?
print(my_kitchen >17 or my_kitchen< 14)

# Double my_kitchen smaller than triple your_kitchen?
print(my_kitchen * 2 < your_kitchen * 3)
```

```python
# Create arrays
import numpy as np
my_house = np.array([18.0, 20.0, 10.75, 9.50])
your_house = np.array([14.0, 24.0, 14.25, 9.0])

# my_house greater than 18.5 or smaller than 10
print(np.logical_or(my_house>18.5, my_house<10))

# Both my_house and your_house smaller than 11
print(np.logical_and(my_house<11, your_house<11))
```

```python
# NumPy is imported, seed is set

import numpy as np
np.random.seed(123)

# Starting step
step = 50

# Roll the dice
dice = np.random.randint(1, 7)

# Finish the control construct
if dice <= 2:
    step = step - 1
elif dice <= 5:
    step = step
else:
    step = step + np.random.randint(1, 7)

# Print out dice and step
print("Dice:", dice)
print("Step:", step)
```

```python
random_walk = [0]

for x in range(100):
    # Set step: last element in random_walk
    step = random_walk[-1]

    # Roll the dice
    dice = np.random.randint(1, 7)

    # Determine next step
    if dice <= 2:
        step = step - 1
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1, 7)

    # append next_step to random_walk
    random_walk.append(step)

# Print random_walk
print(random_walk)
```

```python
# NumPy is imported, seed is set
import numpy as np

# Initialize random_walk
random_walk = [0]

for x in range(100):
    step = random_walk[-1]
    dice = np.random.randint(1, 7)

    if dice <= 2:
        # Make sure step can't go below 0
        step = max(0, step - 1)
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1, 7)

    random_walk.append(step)

print(random_walk)
```

```python
# NumPy is imported; seed is set
import numpy as np

# Initialize all_walks (don't change this line)
all_walks = []

# Simulate random walk five times
for i in range(5):

    # Code from before
    random_walk = [0]

    for x in range(100):
        step = random_walk[-1]
        dice = np.random.randint(1, 7)

        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1, 7)

        random_walk.append(step)

    # Append random_walk to all_walks
    all_walks.append(random_walk)

# Print all_walks
print(all_walks)
```
