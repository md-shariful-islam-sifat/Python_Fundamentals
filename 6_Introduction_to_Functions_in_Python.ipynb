```python
# Define shout with parameters word1 and word2
def shout(word1, word2):
    """Concatenate strings with three exclamation marks"""
    # Concatenate word1 with '!!!': shout1
    shout1 = word1 + '!!!'

    # Concatenate word2 with '!!!': shout2
    shout2 = word2 + '!!!'

    # Concatenate shout1 with shout2: new_shout
    new_shout= shout1+  shout2

    # Return new_shout
    return new_shout

# Pass 'congratulations' and 'you' to shout(): yell
yell = shout('congratulations', 'you')

# Print yell
print(yell)
```

```python
# Define shout_all with parameters word1 and word2
def shout_all(word1, word2):
    """Return a tuple of strings"""

    # Concatenate word1 with '!!!': shout1
    shout1 = word1 + "!!!"

    # Concatenate word2 with '!!!': shout2
    shout2 = word2 + "!!!"

    # Construct a tuple with shout1 and shout2
    shout_words = (shout1, shout2)

    # Return shout_words
    return shout_words


# Pass 'congratulations' and 'you' to shout_all()
yell1, yell2 = shout_all("congratulations", "you")

# Print yell1 and yell2
print(yell1)
print(yell2)
```

```python
import pandas as pd

# Create Employee DataFrame
df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sakib", "Nadia", "Ayesha", "Hasan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [50000, 40000, 60000, 55000, 45000, 65000]
})


# Define count_entries()
def count_entries(df, col_name):
    """Return a dictionary with counts of occurrences as value for each key."""

    # Initialize an empty dictionary
    langs_count = {}

    # Extract column from DataFrame
    col = df[col_name]

    # Iterate over column
    for entry in col:

        # If the entry is already in dictionary, add 1
        if entry in langs_count.keys():
            langs_count[entry] += 1

        # Otherwise, add the entry with value 1
        else:
            langs_count[entry] = 1

    # Return the dictionary
    return langs_count


# Call count_entries()
result = count_entries(df, "Department")

# Print the result
print(result)
```

```python
import pandas as pd

# Create Employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Nadia"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 40000, 55000]
})


# Define employee_info
def employee_info(name1, name2, name3):
    """Return employee names with ' - Employee'."""

    # Define nested function
    def inner(name):
        """Add ' - Employee' to the name."""
        return name + " - Employee"

    # Return a tuple
    return (inner(name1), inner(name2), inner(name3))


# Get employee names from DataFrame
name1 = employees.loc[0, "Name"]
name2 = employees.loc[1, "Name"]
name3 = employees.loc[2, "Name"]

# Call the function
result = employee_info(name1, name2, name3)

# Print result
print(result)
```

```python
import pandas as pd

# Create Employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Nadia"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 40000, 55000]
})


# Define employee_repeat
def employee_repeat(n):
    """Return the inner function."""

    # Define inner function
    def inner_employee(name):
        """Repeat employee name n times."""
        employee_name = name + " "
        return employee_name * n

    # Return inner function
    return inner_employee


# Create functions
twice = employee_repeat(2)
thrice = employee_repeat(3)


# Get employee name
name = employees.loc[0, "Name"]

# Call the functions
print(twice(name))
print(thrice(name))
```

```python
# Define employee_status()
def employee_status(name):
    """Change an employee's status."""

    # Create a variable in the outer function
    status = "Intern"

    # Print the original status
    print(name, "is", status)

    # Define nested function
    def promote():
        """Change the status variable."""

        # Use status from the enclosing function
        nonlocal status

        # Change the status
        status = "Full-Time Employee"

    # Call the nested function
    promote()

    # Print the updated status
    print(name, "is now", status)


# Call the function
employee_status("Rahim")
```

```python
import pandas as pd

# Create Employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Nadia", "Hasan"],
    "Department": ["IT", "HR", "Finance", "IT"],
    "Salary": [50000, 40000, 55000, 60000]
})


# Define employee_names()
def employee_names(*args):
    """Concatenate employee names together."""

    # Initialize an empty string
    names = ""

    # Concatenate the employee names
    for name in args:
        names += name + " "

    # Return names
    return names.strip()


# Call with one employee
one_employee = employee_names("Rahim")

# Call with multiple employees
many_employees = employee_names("Rahim", "Karim", "Nadia", "Hasan")

# Print results
print(one_employee)
print(many_employees)
```

```python
import pandas as pd

# Employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Nadia"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 40000, 55000]
})


# Define employee_message
def employee_message(name, repeat=1):
    """Repeat an employee name and add a message."""

    # Initialize empty strings
    employee_name = ""
    message = ""

    # Exception handling
    try:
        # Repeat the employee name
        employee_name = name * repeat

        # Add message
        message = employee_name + " - Employee"

    except TypeError:
        print("Name must be a string and repeat must be an integer.")

    # Return message
    return message


# Correct call
print(employee_message("Rahim", 3))

# Incorrect call
print(employee_message("Karim", "three"))
```

```python
def employee_message(name, repeat=1):
    """Repeat an employee name."""

    if repeat < 0:
        raise ValueError("repeat must be a non-negative integer.")

    employee_name = name * repeat
    return employee_name + " - Employee"


print(employee_message("Rahim", 3))
```

```python
import pandas as pd

# Create Employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Nadia", "Hasan", "Ayesha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Status": ["Active", "Active", "On Leave", "Active", "Active"]
})


# Define count_entries()
def count_entries(df, col_name="Department"):
    """Return a dictionary with counts of occurrences."""

    # Initialize an empty dictionary
    cols_count = {}

    # Add try block
    try:
        # Extract column from DataFrame
        col = df[col_name]

        # Iterate over the column
        for entry in col:

            # If entry is already in dictionary, add 1
            if entry in cols_count:
                cols_count[entry] += 1

            # Otherwise, add the entry with value 1
            else:
                cols_count[entry] = 1

        # Return the dictionary
        return cols_count

    # Handle missing column
    except KeyError:
        print("The DataFrame does not have the requested column.")


# Call count_entries() with Department
result1 = count_entries(employees, "Department")

# Print result
print(result1)
```

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Nadia", "Hasan"],
    "Department": ["IT", "HR", "IT", "Finance"]
})


def count_entries(df, col_name="Department"):

    if col_name not in df.columns:
        raise ValueError("Column is not in DataFrame")

    cols_count = {}

    for entry in df[col_name]:
        if entry in cols_count:
            cols_count[entry] += 1
        else:
            cols_count[entry] = 1

    return cols_count


result1 = count_entries(employees, "Department")

print(result1)
```
