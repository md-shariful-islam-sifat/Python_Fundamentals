```python
# Import the matplotlib.pyplot submodule and name it plt
import matplotlib.pyplot as plt

# Create a Figure and an Axes with plt.subplots
fig, ax = plt.subplots()

# Call the show function to show the result
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

seattle = pd.DataFrame({
    "MONTH": [1, 2, 3, 4, 5, 6],
    "Rainfall": [150, 120, 100, 80, 60, 40]
})

austin = pd.DataFrame({
    "MONTH": [1, 2, 3, 4, 5, 6],
    "Rainfall": [70, 60, 50, 40, 30, 20]
})

fig, ax = plt.subplots()   #Create Figure and Axes

ax.plot(seattle["MONTH"], seattle["Rainfall"], label="Seattle")
ax.plot(austin["MONTH"], austin["Rainfall"], label="Austin")

ax.set_xlabel("Month")
ax.set_ylabel("Rainfall")
ax.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

seattle_weather = pd.DataFrame({
    "MONTH": [1, 2, 3, 4, 5, 6],
    "Rainfall": [150, 120, 100, 80, 60, 40]
})

austin_weather = pd.DataFrame({
    "MONTH": [1, 2, 3, 4, 5, 6],
    "Rainfall": [70, 60, 50, 40, 30, 20]
})

fig, ax = plt.subplots()


# Plot Seattle data, setting data appearance
ax.plot(seattle_weather["MONTH"], seattle_weather["Rainfall"], color='b', marker='o', linestyle='--')

# Plot Austin data, setting data appearance
ax.plot(austin_weather["MONTH"], austin_weather["Rainfall"], color='r', marker='v', linestyle='--')

# Call show to display the resulting plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create Seattle DataFrame
seattle_weather = pd.DataFrame({
    "MONTH": [1, 2, 3, 4, 5, 6],
    "MLY-PRCP-NORMAL": [150, 120, 100, 80, 60, 40],
    "MLY-PRCP-25PCTL": [100, 80, 70, 50, 40, 25],
    "MLY-PRCP-75PCTL": [200, 160, 140, 110, 80, 60]
})

# Create Austin DataFrame
austin_weather = pd.DataFrame({
    "MONTH": [1, 2, 3, 4, 5, 6],
    "MLY-PRCP-NORMAL": [70, 60, 50, 40, 30, 20],
    "MLY-PRCP-25PCTL": [40, 35, 30, 25, 20, 10],
    "MLY-PRCP-75PCTL": [100, 90, 75, 60, 45, 30]
})


# Create a figure and an array of axes: 2 rows, 1 column with shared y axis

fig, ax = plt.subplots(2, 1, sharey=True)

# Plot Seattle precipitation data in the top axes

ax[0].plot(seattle_weather["MONTH"], seattle_weather["MLY-PRCP-NORMAL"], color="b")
ax[0].plot(seattle_weather["MONTH"], seattle_weather["MLY-PRCP-25PCTL"], color="b", linestyle="--")
ax[0].plot(seattle_weather["MONTH"], seattle_weather["MLY-PRCP-75PCTL"], color="b", linestyle=":")

# Plot Austin precipitation data in the bottom axes

ax[1].plot(austin_weather["MONTH"], austin_weather["MLY-PRCP-NORMAL"], color="r")
ax[1].plot(austin_weather["MONTH"], austin_weather["MLY-PRCP-25PCTL"], color="r", linestyle="--")
ax[1].plot(austin_weather["MONTH"], austin_weather["MLY-PRCP-75PCTL"], color="r", linestyle=":")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a DataFrame
climate_change = pd.DataFrame({
    "CO2": [315, 320, 330, 340, 350, 360],
    "relative_temp": [0.1, 0.2, 0.3, 0.5, 0.7, 0.9]
}, index=[1960, 1970, 1980, 1990, 2000, 2010])


# Initalize a Figure and Axes

fig, ax = plt.subplots()


# Plot the CO2 variable in blue

ax.plot(climate_change.index,
        climate_change["CO2"],
        color="blue")


# Create a twin Axes that shares the x-axis

ax2 = ax.twinx()


# Plot the relative temperature in red

ax2.plot(climate_change.index,
         climate_change["relative_temp"],
         color="red")


# Show the plot

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a DataFrame
climate_change = pd.DataFrame({
    "CO2": [315, 320, 330, 340, 350, 360],
    "relative_temp": [0.1, 0.2, 0.3, 0.5, 0.7, 0.9]
}, index=[1960, 1970, 1980, 1990, 2000, 2010])


# Define the function
def plot_timeseries(axes, x, y, color, xlabel, ylabel):

    # Plot the inputs x and y
    axes.plot(x, y, color=color)

    # Set the x-axis label
    axes.set_xlabel(xlabel)

    # Set the y-axis label
    axes.set_ylabel(ylabel, color=color)

    # Set the color of y-axis ticks
    axes.tick_params("y", colors=color)


# Create Figure and Axes
fig, ax = plt.subplots()


# Use the function to plot CO2
plot_timeseries(
    ax,
    climate_change.index,
    climate_change["CO2"],
    "blue",
    "Year",
    "CO2 Concentration"
)


plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a DataFrame
climate_change = pd.DataFrame({
    "co2": [315, 320, 330, 340, 350, 360],
    "relative_temp": [0.10, 0.20, 0.30, 0.50, 0.70, 0.90]
}, index=[1960, 1970, 1980, 1990, 2000, 2010])


# Define the function
def plot_timeseries(axes, x, y, color, xlabel, ylabel):

    # Plot x and y
    axes.plot(x, y, color=color)

    # Set x-axis label
    axes.set_xlabel(xlabel)

    # Set y-axis label
    axes.set_ylabel(ylabel, color=color)

    # Set y-axis tick color
    axes.tick_params("y", colors=color)


# Create Figure and Axes
fig, ax = plt.subplots()


# Plot CO2 levels in blue
plot_timeseries(
    ax,
    climate_change.index,
    climate_change["co2"],
    "blue",
    "Time (years)",
    "CO2 levels"
)


# Create a twin Axes
ax2 = ax.twinx()


# Plot relative temperature in red
plot_timeseries(
    ax2,
    climate_change.index,
    climate_change["relative_temp"],
    "red",
    "Time (years)",
    "Relative temperature (Celsius)"
)


# Show the plot
plt.show()
```

**Bar Chart**

```python
import pandas as pd
import matplotlib.pyplot as plt

medals = pd.DataFrame({
    "Gold": [10, 8, 6],
    "Silver": [5, 7, 4],
    "Bronze": [3, 6, 5]
}, index=["USA", "China", "Japan"])

fig, ax = plt.subplots()

# Add bars for "Gold" with the label "Gold"
ax.bar(medals.index, medals["Gold"], label="Gold")

# Stack bars for "Silver" on top with label "Silver"
ax.bar(
    medals.index,
    medals["Silver"],
    bottom=medals["Gold"],
    label="Silver"
)

# Stack bars for "Bronze" on top of that with label "Bronze"
ax.bar(
    medals.index,
    medals["Bronze"],
    bottom=medals["Gold"] + medals["Silver"],
    label="Bronze"
)

# Display the legend
ax.legend()

plt.show()
```

**Histogram**

```python
import pandas as pd
import matplotlib.pyplot as plt

mens_rowing = pd.DataFrame({
    "Weight": [80, 82, 85, 88, 90, 92, 95, 98, 100]
})

mens_gymnastics = pd.DataFrame({
    "Weight": [55, 58, 60, 62, 65, 67, 70, 72]
})

fig, ax = plt.subplots()

# Plot a histogram of "Weight" for mens_rowing
ax.hist(mens_rowing["Weight"])

# Compare to histogram of "Weight" for mens_gymnastics
ax.hist(mens_gymnastics["Weight"])

# Set the x-axis label to "Weight (kg)"
ax.set_xlabel("Weight (kg)")

# Set the y-axis label to "# of observations"
ax.set_ylabel("# of observations")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create DataFrame for men's rowing
mens_rowing = pd.DataFrame({
    "Weight": [80, 82, 85, 87, 88, 90, 92, 94, 96, 98]
})

# Create DataFrame for men's gymnastics
mens_gymnastics = pd.DataFrame({
    "Weight": [55, 57, 60, 62, 64, 66, 68, 70, 72, 75]
})


fig, ax = plt.subplots()

# Plot a histogram of "Weight" for mens_rowing
ax.hist(
    mens_rowing["Weight"],
    label="Rowing",
    histtype="step",
    bins=5
)

# Compare to histogram of "Weight" for mens_gymnastics
ax.hist(
    mens_gymnastics["Weight"],
    label="Gymnastics",
    histtype="step",
    bins=5
)

# Set the x-axis label
ax.set_xlabel("Weight (kg)")

# Set the y-axis label
ax.set_ylabel("# of observations")

# Add the legend
ax.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

mens_rowing = pd.DataFrame({
    "Height": [190, 192, 188, 195, 191, 189]
})

mens_gymnastics = pd.DataFrame({
    "Height": [165, 168, 162, 170, 166, 164]
})

fig, ax = plt.subplots()

ax.bar(
    "Rowing",
    mens_rowing["Height"].mean(),
    yerr=mens_rowing["Height"].std()
)

ax.bar(
    "Gymnastics",
    mens_gymnastics["Height"].mean(),
    yerr=mens_gymnastics["Height"].std()
)

ax.set_ylabel("Height (cm)")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create Seattle weather DataFrame
seattle_weather = pd.DataFrame({
    "MONTH": range(1, 13),
    "MLY-TAVG-NORMAL": [
        42, 44, 47, 50, 55, 60,
        65, 66, 60, 53, 46, 42
    ],
    "MLY-TAVG-STDDEV": [
        3, 3, 4, 4, 4, 4,
        3, 3, 4, 4, 4, 3
    ]
})

# Create Austin weather DataFrame
austin_weather = pd.DataFrame({
    "MONTH": range(1, 13),
    "MLY-TAVG-NORMAL": [
        50, 54, 62, 70, 77, 83,
        85, 86, 80, 71, 61, 52
    ],
    "MLY-TAVG-STDDEV": [
        4, 4, 5, 5, 5, 4,
        4, 4, 5, 5, 5, 4
    ]
})


# Create Figure and Axes
fig, ax = plt.subplots(figsize=(10, 6))

# Add Seattle temperature data with error bars
ax.errorbar(
    seattle_weather["MONTH"],
    seattle_weather["MLY-TAVG-NORMAL"],
    yerr=seattle_weather["MLY-TAVG-STDDEV"],
    label="Seattle",
    marker="o"
)

# Add Austin temperature data with error bars
ax.errorbar(
    austin_weather["MONTH"],
    austin_weather["MLY-TAVG-NORMAL"],
    yerr=austin_weather["MLY-TAVG-STDDEV"],
    label="Austin",
    marker="s"
)

# Set labels
ax.set_xlabel("Month")
ax.set_ylabel("Temperature (Fahrenheit)")

# Set x-axis ticks
ax.set_xticks(range(1, 13))

# Add legend
ax.legend()

# Add grid
ax.grid(True, alpha=0.3)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a sample DataFrame
summer_2016_medals = pd.DataFrame({
    "Sport": [
        "Athletics", "Athletics", "Athletics",
        "Swimming", "Swimming", "Swimming",
        "Cycling", "Cycling", "Cycling",
        "Gymnastics", "Gymnastics", "Gymnastics"
    ],
    "Weight": [
        75, 80, 78,
        70, 72, 68,
        73, 76, 71,
        55, 58, 54
    ]
})

# Find unique sports
sports = summer_2016_medals["Sport"].unique()

fig, ax = plt.subplots()

# Loop over the different sports
for sport in sports:

    # Extract rows for this sport
    sport_df = summer_2016_medals[
        summer_2016_medals["Sport"] == sport
    ]

    # Add bar for mean Weight with standard deviation error bar
    ax.bar(
        sport,
        sport_df["Weight"].mean(),
        yerr=sport_df["Weight"].std()
    )

ax.set_ylabel("Weight")
ax.set_xticklabels(sports, rotation=90)

# Save the figure
fig.savefig("sports_weights.png")

plt.show()
```
