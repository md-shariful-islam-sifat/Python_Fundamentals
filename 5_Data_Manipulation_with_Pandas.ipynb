```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sakib", "Nadia", "Ayesha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 35000, 60000, 55000, 40000],
    "Experience": [2, 1, 5, 4, 3]
})

# Create salary_per_10_years col
employees["salary_per_10_years"] = (
    10 * employees["Salary"] / employees["Experience"]
)

# Subset rows for salary_per_10_years greater than 100000
high_salary = employees[
    employees["salary_per_10_years"] > 100000
]

# Sort high_salary by descending salary_per_10_years
high_salary_srt = high_salary.sort_values(
    "salary_per_10_years",
    ascending=False
)

# Select the Name and salary_per_10_years columns
result = high_salary_srt[
    ["Name", "salary_per_10_years"]
]

# See the result
print(result)
```

```python
import pandas as pd

# Create employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sakib", "Nadia", "Ayesha", "Hasan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Job_Type": ["Full-Time", "Part-Time", "Full-Time",
                 "Full-Time", "Part-Time", "Full-Time"],
    "Salary": [45000, 35000, 60000, 55000, 40000, 50000]
})

# Count employees in each department
dept_counts = employees["Department"].value_counts()
print("Department counts:")
print(dept_counts)

# Get the proportion of employees in each department
dept_props = employees["Department"].value_counts(normalize=True)
print("\nDepartment proportions:")
print(dept_props)

# Count job types and sort
job_counts_sorted = employees["Job_Type"].value_counts(sort=True)
print("\nJob type counts:")
print(job_counts_sorted)

# Get job type proportions and sort
job_props_sorted = employees["Job_Type"].value_counts(
    sort=True,
    normalize=True
)
print("\nJob type proportions:")
print(job_props_sorted)
```

```python
# Calculate total salary
salary_all = employees["Salary"].sum()

# Total salary for IT
salary_IT = employees[employees["Department"] == "IT"]["Salary"].sum()

# Total salary for HR
salary_HR = employees[employees["Department"] == "HR"]["Salary"].sum()

# Total salary for Finance
salary_Finance = employees[
    employees["Department"] == "Finance"
]["Salary"].sum()

# Get proportion for each department
salary_propn_by_dept = [
    salary_IT,
    salary_HR,
    salary_Finance
] / salary_all

print(salary_propn_by_dept)
```

```python
# Group by Department; calculate total Salary
salary_by_dept = employees.groupby("Department")["Salary"].sum()

# Get proportion for each department
salary_propn_by_dept = salary_by_dept / sum(salary_by_dept)

print(salary_propn_by_dept)
```

```python
# Group by Department and Job_Type; calculate total Salary
salary_by_dept_job = employees.groupby(
    ["Department", "Job_Type"]
)["Salary"].sum()

print(salary_by_dept_job)
```

```python
salary_stats = employees.groupby("Department")["Salary"].agg(["min", "max", "mean", "median"])

print(salary_stats)
```

```python
mean_salary_by_dept = employees.pivot_table(
    values="Salary",
    index="Department"
)

print(mean_salary_by_dept)
```

```python
mean_med_salary_by_dept = employees.pivot_table(
    values="Salary",
    index="Department",
    aggfunc=["mean", "median"]
)

print(mean_med_salary_by_dept)
```

```python
import pandas as pd

# Create employee DataFrame
employees = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sakib", "Nadia", "Ayesha", "Hasan"],
    "Department": ["Finance", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [55000, 40000, 60000, 58000, 42000, 65000],
    "Hire_Date": [
        "2020-03-15",
        "2021-06-20",
        "2022-01-10",
        "2021-08-05",
        "2022-04-12",
        "2020-11-25"
    ]
})

# Convert Hire_Date to datetime
employees["Hire_Date"] = pd.to_datetime(employees["Hire_Date"])

# Add year column
employees["year"] = employees["Hire_Date"].dt.year

# Create pivot table
salary_by_dept_vs_year = employees.pivot_table(
    values="Salary",
    index="Department",
    columns="year"
)

print(salary_by_dept_vs_year)
```

```python
# Average salary for each year
mean_salary_by_year = salary_by_dept_vs_year.mean()

# Find the year with the highest average salary
print(
    mean_salary_by_year[
        mean_salary_by_year == mean_salary_by_year.max()
    ]
)

# Average salary for each department
mean_salary_by_dept = salary_by_dept_vs_year.mean(axis="columns")

# Find the department with the lowest average salary
print(
    mean_salary_by_dept[
        mean_salary_by_dept == mean_salary_by_dept.min()
    ]
)
```
