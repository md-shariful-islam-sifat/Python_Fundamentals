**Mathematical operations**

```python
a = 5 #integer
b = 6 #integer

# Addition
print("a+b = ", a+b)

# Multiplication
print("a*b = ", a*b)

# Subtraction
print("a-b = ", a-b)

# Modulus: Divides left hand operand by right hand operand and returns remainder
print("a%b = ", a%b)

# Division
print("a/b = ", a/b)

# Floor Division - The division of operands
# where the result is the quotient in which
# the digits after the decimal point are removed.
# But if one of the operands is negative, the result is floored,
# i.e., rounded away from zero (towards negative infinity)
print("a//b = ", a//b)

# Exponent: Performs exponential (power) calculation on operators
print("a**b = ", a**b)
```

**NumPy** <br>
NumPy (Numerical Python) is a Python library used for fast numerical computations. It provides a powerful data structure called the NumPy array (ndarray), which is more efficient than Python lists for mathematical operations.






*   array[i] → One element.
*   array[start:end] → Elements from start to end-1.
*   array[:5] → First 5 elements.
*   array[5:] → From index 5 to the end.
*   array[-1] → Last element.



```python
import pandas as pd
mlb = pd.read_csv("https://assets.datacamp.com/course/intro_to_python/baseball.csv")
height_in = mlb['Height'].tolist()  # Make sure to use this list

import numpy as np
# Create a numpy array from height_in
np_height_in = np.array(height_in)

# Print out np_height_in
print(np_height_in)

# Convert np_height_in to meters
np_height_m = np_height_in * 0.0254

# Print np_height_m
print(np_height_m)
```

```python
import numpy as np

baseball = [[180, 78.4],
            [215, 102.7],
            [210, 98.5],
            [188, 75.2]]

# Create a 2D NumPy array
np_baseball = np.array(baseball)

# Print the first row (instead of the 50th row, since there are only 4 rows)
print("First row:")
print(np_baseball[0])

# Print the second row
print("\nSecond row:")
print(np_baseball[1])

# Select the entire second column (weights)
np_weight_lb = np_baseball[:, 1]

print("\nWeights:")
print(np_weight_lb)

# Print the height of the fourth player
print("\nHeight of the fourth player:")
print(np_baseball[3, 0])
```

```python
import numpy as np

# Sample baseball dataset
# [Height (inches), Weight (pounds)]
baseball = [
    [74, 180],
    [72, 215],
    [75, 210],
    [73, 188],
    [76, 220],
    [71, 175]
]

# Convert list to NumPy array
np_baseball = np.array(baseball)

# Select the first column (heights)
np_height_in = np_baseball[:, 0]

# Print the heights
print("Heights (inches):")
print(np_height_in)

# Calculate the mean
print("\nMean Height:")
print(np.mean(np_height_in))

# Calculate the median
print("\nMedian Height:")
print(np.median(np_height_in))
```



*   np_baseball[:, 0] → Selects all rows of the first column (heights).
*   np.mean(np_height_in) → Calculates the average height.
*   np.median(np_height_in) → Calculates the median height



```python
import numpy as np

# Height (inches), Weight (pounds)
baseball = [
    [74, 180],
    [72, 215],
    [75, 210],
    [73, 188],
    [76, 220],
    [71, 175]
]

np_baseball = np.array(baseball)

avg = np.mean(np_baseball[:, 0])
print("Average:", avg)

med = np.median(np_baseball[:, 0])
print("Median:", med)

stddev = np.std(np_baseball[:, 0])
print("Standard Deviation:", stddev)

corr = np.corrcoef(np_baseball[:, 0], np_baseball[:, 1])
print("Correlation Matrix:\n", corr)3
```
